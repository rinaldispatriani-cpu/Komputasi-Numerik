```python

# Mengimpor library NumPy dengan nama alias np.
# NumPy digunakan untuk membantu perhitungan matematika,
# terutama operasi matriks dan array.
import numpy as np
```

```python

# Membuat fungsi bernama eliminasi_gauss.
# A_matrix berisi matriks koefisien persamaan.
# b_vector berisi nilai konstanta dari setiap persamaan.
def eliminasi_gauss(A_matrix, b_vector):

    # Menghitung jumlah persamaan berdasarkan panjang b_vector.
    # Nilai n juga menunjukkan jumlah variabel jika sistem berbentuk persegi.
    n = len(b_vector)

    # Mengubah matriks koefisien menjadi tipe float
    # agar dapat digunakan dalam perhitungan desimal.
    #
    # b_vector.reshape(-1, 1) mengubah vektor konstanta
    # menjadi matriks satu kolom.
    #
    # np.hstack menggabungkan matriks A dan vektor b
    # menjadi matriks augmented berukuran n x (n+1).
    M = np.hstack([
        A_matrix.astype(float),
        b_vector.reshape(-1, 1).astype(float)
    ])

    # Memulai proses eliminasi Gauss.
    # Perulangan dilakukan untuk setiap kolom pivot.
    for kolom in range(n):

        # Mencari baris pivot berdasarkan nilai absolut terbesar
        # pada kolom yang sedang diproses.
        # Tujuannya agar perhitungan lebih stabil.
        baris_pivot = kolom + np.argmax(
            np.abs(M[kolom:n, kolom])
        )

        # Memeriksa apakah posisi baris pivot berbeda
        # dari baris yang sedang diproses.
        if baris_pivot != kolom:

            # Menukar posisi kedua baris apabila diperlukan.
            # Baris dengan nilai pivot terbesar ditempatkan
            # pada posisi baris aktif.
            M[[kolom, baris_pivot]] = M[[baris_pivot, kolom]]

        # Memeriksa apakah nilai pivot terlalu dekat dengan nol.
        # Jika iya, program menghentikan perhitungan karena
        # sistem tidak memiliki solusi tunggal melalui prosedur ini.
        if np.abs(M[kolom, kolom]) < 1e-9:
            raise ValueError(
                "Sistem tidak memiliki solusi tunggal "
                "(pivot mendekati nol)."
            )

        # Melakukan eliminasi pada semua baris
        # yang berada di bawah baris pivot.
        for i in range(kolom + 1, n):

            # Menghitung faktor eliminasi.
            # Faktor diperoleh dari nilai pada kolom aktif
            # dibagi dengan nilai pivot.
            faktor = M[i, kolom] / M[kolom, kolom]

            # Mengurangi baris saat ini dengan hasil perkalian
            # faktor dan baris pivot.
            # Operasi ini bertujuan membuat nilai di bawah pivot
            # menjadi nol.
            M[i, kolom:] -= faktor * M[kolom, kolom:]

    # Memulai proses substitusi balik (back-substitution).

    # Membuat array x berisi nilai awal nol.
    # Array ini digunakan untuk menyimpan hasil variabel.
    x = np.zeros(n)

    # Melakukan perulangan dari baris terakhir menuju baris pertama.
    # Karena matriks sudah berbentuk segitiga atas,
    # variabel dihitung dari baris paling bawah.
    for i in range(n - 1, -1, -1):

        # Menghitung jumlah perkalian koefisien
        # dengan nilai variabel yang sudah diketahui.
        # np.dot digunakan untuk perkalian dan penjumlahan array.
        jumlah = np.dot(M[i, i+1:n], x[i+1:n])

        # Menghitung nilai variabel pada posisi ke-i.
        # Nilai konstanta dikurangi jumlah hasil perkalian,
        # kemudian dibagi dengan koefisien diagonal.
        x[i] = (M[i, n] - jumlah) / M[i, i]

    # Mengembalikan seluruh nilai variabel yang telah dihitung.
    return x
```

```python

# Menentukan matriks koefisien dari tiga persamaan linear.
#
# Persamaan 1:  2x + y - z = 8
# Persamaan 2: -3x - y + 2z = -11
# Persamaan 3: -2x + y + 2z = -3

A = np.array([
    [2, 1, -1],    # Koefisien x, y, dan z pada persamaan pertama.
    [-3, -1, 2],   # Koefisien x, y, dan z pada persamaan kedua.
    [-2, 1, 2]     # Koefisien x, y, dan z pada persamaan ketiga.
])

# Menentukan vektor konstanta dari ketiga persamaan.
# Angka 8, -11, dan -3 merupakan nilai di sebelah kanan
# tanda sama dengan pada masing-masing persamaan.
b = np.array([8, -11, -3])
```

```python

# Menggunakan try untuk menjalankan fungsi eliminasi Gauss.
try:

    # Memanggil fungsi eliminasi_gauss dengan memasukkan
    # matriks koefisien A dan vektor konstanta b.
    hasil = eliminasi_gauss(A, b)

    # Menampilkan hasil perhitungan variabel x, y, dan z.
    # :.4f berarti setiap nilai ditampilkan dengan
    # empat angka di belakang koma.
    print(
        f"Solusi: x = {hasil[0]:.4f}, "
        f"y = {hasil[1]:.4f}, "
        f"z = {hasil[2]:.4f}"
    )

# Menangkap kesalahan ValueError jika terjadi
# masalah pada nilai pivot.
except ValueError as e:

    # Menampilkan pesan kesalahan yang diberikan fungsi.
    print("Error:", e)
```
