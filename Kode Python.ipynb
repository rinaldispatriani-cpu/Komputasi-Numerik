```python
import numpy as np

def eliminasi_gauss(A_matrix, b_vector):
    """
    Menyelesaikan sistem persamaan linear Ax = b menggunakan eliminasi Gauss dengan pivot sebagian.
    """
    n = len(b_vector)
    # Bentuk matriks augmented A berukuran n x (n+1)
    M = np.hstack([A_matrix.astype(float), b_vector.reshape(-1, 1).astype(float)])

    # Proses Eliminasi
    for kolom in range(n):
        # Tentukan baris pivot dengan nilai absolut terbesar pada kolom aktif
        baris_pivot = kolom + np.argmax(np.abs(M[kolom:n, kolom]))

        # Tukar baris aktif dengan baris pivot
        if baris_pivot != kolom:
            M[[kolom, baris_pivot]] = M[[baris_pivot, kolom]]

        # JIKA nilai pivot mendekati nol, periksa solusi
        if np.abs(M[kolom, kolom]) < 1e-9:
            raise ValueError("Sistem tidak memiliki solusi tunggal (pivot mendekati nol).")

        # UNTUK setiap baris di bawah baris aktif
        for i in range(kolom + 1, n):
            faktor = M[i, kolom] / M[kolom, kolom]
            M[i, kolom:] -= faktor * M[kolom, kolom:]

    # Substitusi Balik (Back-substitution)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        jumlah = np.dot(M[i, i+1:n], x[i+1:n])
        x[i] = (M[i, n] - jumlah) / M[i, i]

    return x

# Membaca/menentukan koefisien persamaan
# Persamaan 1: 2x + y - z = 8
# Persamaan 2: -3x - y + 2z = -11
# Persamaan 3: -2x + y + 2z = -3
A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
])
b = np.array([8, -11, -3])

try:
    hasil = eliminasi_gauss(A, b)
    print(f"Solusi: x = {hasil[0]:.4f}, y = {hasil[1]:.4f}, z = {hasil[2]:.4f}")
except ValueError as e:
    print("Error:", e)
```
